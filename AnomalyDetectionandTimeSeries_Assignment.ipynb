# Anomaly Detection and Time Series Assignment


1.	What is Anomaly Detection? Explain its types (point, contextual, and collective anomalies) with examples.
    -	Anomaly detection is the process of identifying data points, events, or observations that deviate significantly from a dataset’s normal or expected behavior
Types of Anomalies
Point Anomalies
A point anomaly occurs when a single, isolated data point stands out as entirely abnormal compared to the rest of the dataset. This is the simplest and most common type of anomaly.
Example: A credit card transaction of ₹5,00,000 on a account that typically only spends      between ₹500 and ₹5,000 per transaction. The single value itself is instantly flagged as an outlier.
Contextual Anomalies
A contextual anomaly (also known as a conditional anomaly) happens when a data point is considered abnormal only within a specific context or circumstance. The exact same data point might be completely normal under a different context.
Example: A temperature reading of 35°C (95°F) is completely normal during the summer. However, if that same temperature is recorded in the middle of winter in a cold climate, it is a contextual anomaly.
Collective Anomalies
A collective anomaly occurs when a sequence or collection of data points looks normal individually, but their appearance together as a group indicates an anomaly.
Example: In an online system, a user typing a single character, hitting backspace, or copying text is perfectly normal behavior. However, if a machine records that exact sequence of actions repeated 1,000 times in 10 seconds, the collective sequence indicates a automated bot attack.

3.	What are the key components of a Time Series? Explain each with one example.
    -	A time series is a sequence of data points recorded at consistent time intervals. To analyze and forecast this data, it is broken down into four key components: trend, seasonality, cyclicity, and irregularity (noise).
Trend (𝑇𝑡) The trend represents the long-term, smooth progression of the data over a long period. It shows whether the data is generally moving upward, downward, or remaining stagnant, ignoring short-term fluctuations.
Example: The steady increase in global e-commerce sales over the last decade due to growing internet adoption.
Seasonality (𝑆𝑡)
Seasonality refers to regular, predictable fluctuations that repeat within a specific fixed period, such as a day, week, month, or year. These patterns are usually tied to calendar events or weather seasons.
Example: A sharp spike in ice cream sales every summer, which drops predictably during the winter months year after year.
Cyclicity (𝐶𝑡)
Cyclicity involves long-term oscillations or swings around the trend line that do not have a fixed, predictable period. These cycles typically last longer than a year and are often driven by economic, political, or business factors.
Example: The housing market boom and bust cycles, which mirror multi-year economic recessions and expansions.
Irregularity or Noise (𝐼𝑡)
Irregularity (also called residual or noise) represents the random, unpredictable variations in the data. These fluctuations remain after removing the trend, seasonality, and cycles, and are usually caused by unexpected, short-term events.
Example: A sudden drop in flight bookings during a specific week due to an unexpected volcanic eruption grounding flights.

4.	Define Stationary in time series. How can you test and transform a non-stationary series into a stationary one?
    -	A stationary time series is a data sequence whose statistical properties like mean, variance, and autocorrelation do not change over time.
Stationary data makes it easier for models to learn patterns and make accurate future predictions because the rules governing the past apply equally to the future.
How to Test for Stationarity
You can check if a time series is stationary using visual tools or formal statistical tests.
Visual Inspection: Plot the data and check if the rolling mean or rolling variance changes over time. If the graph trends upward or shows expanding swings, the series is non-stationary.
Augmented Dickey-Fuller (ADF) Test: A statistical test where the null hypothesis assumes the series is non-stationary (has a unit root). If the resulting p-value is less than 0.05, you reject the null hypothesis and confirm the series is stationary.
KPSS Test: A complementary test where the null hypothesis assumes the series is stationary. Using ADF and KPSS together gives a reliable confirmation of stationarity.
How to Transform a Non-Stationary Series
If your data is non-stationary, you can apply mathematical transformations to stabilize the mean and variance.
Differencing (Stabilizes the Mean): Subtract the previous observation from the current observation (yt- yt- 1)  to remove trends. If a first difference does not work, you can apply a second-order difference.
Log Transformation (Stabilizes the Variance): Apply the natural logarithm to the data to reduce the magnitude of growing fluctuations and make the variance constant over time.
Seasonal Differencing (Removes Seasonality): Subtract the observation from the same season in the previous year(yt- yt- s) to eliminate regular seasonal spikes or drops.

5.	Differentiate between AR, MA, ARIMA, SARIMA, and SARIMAX models in terms of structure and application.
    -	AR (Autoregressive) Model
Structure: Predicts the future value using a linear combination of its own past values(p)
Application: Used for stationary data where the current state depends directly on past states.
Example: Forecasting the price of a stable commodity based solely on its prices from the last three days.
MA (Moving Average) Model
Structure: Predicts the future value using a linear combination of past forecast errors or random shocks(q)
Application: Used for stationary data that experiences short-lived shocks that adjust back to a mean value over time.
Example: Modeling the rapid stabilization of water pressure in a pipe after a sudden mechanical valve adjustment.
ARIMA (Autoregressive Integrated Moving Average)
Structure: Combines AR(p) and MA(q) with a differencing component(d) to handle non-stationary data.
Application: Used for non-stationary data that exhibits a clear linear trend but does not have seasonal patterns.
Example: Forecasting national GDP growth or stock market prices that trend upward over years.
Structure: Extends ARIMA by adding dedicated seasonal autoregressive, moving average, and differencing parameters denoted as
 	SARIMA(p,d,q)*(P,D,Q)
 	Application: Used for non-stationary data that exhibits both clear long-term trends and repeating seasonal cycles (e.g., daily, monthly, or yearly).
 	Example: Forecasting monthly airline passenger traffic, which has an upward trend over decades but spikes every summer holiday season.
 	SARIMAX (Seasonal ARIMA with Exogenous Regressors)
 	Structure: Inherits the complete structural framework of SARIMA and introduces exogenous variables(X)  These are external, independent variables that help explain variations in the target time series.
 	Application: Used when the time series has a trend, seasonality, and is also heavily impacted by outside factors or external datasets.
 	Example: Forecasting electricity demand, which depends on internal time factors (trends and seasonality) but is heavily driven by external weather temperatures and holiday schedules.


```python
#Q6. Load a time series dataset (e.g., AirPassengers), plot the original series,
#and decompose it into trend, seasonality, and residual components
#(Include your Python code
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# 1. Complete AirPassengers dataset (Monthly totals from Jan 1949 to Dec 1960)
passengers = [
    112, 118, 132, 129, 121, 135, 148, 148, 136, 119, 104, 118,
    115, 126, 141, 135, 125, 149, 170, 170, 158, 133, 114, 140,
    145, 150, 178, 163, 172, 178, 199, 199, 184, 162, 146, 166,
    171, 180, 193, 181, 183, 218, 230, 242, 209, 191, 172, 194,
    196, 196, 236, 235, 229, 243, 264, 272, 237, 211, 180, 201,
    204, 188, 235, 227, 234, 264, 302, 293, 259, 229, 203, 229,
    242, 233, 267, 269, 270, 315, 364, 347, 312, 274, 237, 278,
    284, 277, 317, 313, 318, 374, 413, 405, 355, 306, 271, 306,
    315, 301, 356, 348, 355, 422, 465, 467, 404, 347, 305, 336,
    340, 318, 362, 348, 363, 435, 491, 505, 404, 359, 310, 337,
    360, 342, 406, 396, 420, 472, 548, 559, 463, 407, 362, 405,
    417, 391, 419, 461, 472, 535, 622, 606, 508, 461, 390, 432
]

# 2. Create date index and build the DataFrame
dates = pd.date_range(start='1949-01-01', end='1960-12-01', freq='MS')
df = pd.DataFrame({'Passengers': passengers}, index=dates)

# 3. Perform Multiplicative Decomposition
# (Multiplicative is used because seasonal variance increases with the trend)
decomposition = seasonal_decompose(df['Passengers'], model='multiplicative', period=12)

# 4. Generate the 4-panel subplot breakdown
fig, axes = plt.subplots(4, 1, figsize=(10, 8), sharex=True)

decomposition.observed.plot(ax=axes[0], color='blue', legend=False)
axes[0].set_ylabel('Observed')
axes[0].set_title('AirPassengers Time Series Decomposition')

decomposition.trend.plot(ax=axes[1], color='orange', legend=False)
axes[1].set_ylabel('Trend')

decomposition.seasonal.plot(ax=axes[2], color='green', legend=False)
axes[2].set_ylabel('Seasonal')

decomposition.resid.plot(ax=axes[3], color='red', style='o', legend=False)
axes[3].set_ylabel('Residuals')

# Add styling and clean layout
for ax in axes:
    ax.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
#Q7. Apply Isolation Forest on a numerical dataset (e.g., NYC Taxi Fare) to
#detect anomalies. Visualize the anomalies on a 2D scatter plot.
#(Include your Python code
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# 1. Synthesize a realistic NYC Taxi Fare dataset with known anomalies
np.random.seed(42)
n_samples = 300

# Normal trips: stable linear relationship between distance and price
distance_normal = np.random.uniform(0.5, 10, n_samples)
fare_normal = distance_normal * 3.5 + np.random.normal(2.5, 2, n_samples)

# Injected anomalies: extreme meter fraud, input bugs, or system glitches
distance_anomalies = np.array([0.1, 0.2, 15.0, 18.0, 1.0, 12.0, 0.5, 14.0, 19.0, 2.0])
fare_anomalies = np.array([75.0, 90.0, 5.0, 12.0, 120.0, 8.0, 85.0, 15.0, 110.0, 95.0])

distances = np.concatenate([distance_normal, distance_anomalies])
fares = np.concatenate([fare_normal, fare_anomalies])
df = pd.DataFrame({'trip_distance': distances, 'fare_amount': fares})

# 2. Configure and fit the Isolation Forest model
# contamination=0.04 explicitly instructs the model to flag roughly 4% of data as outliers
iso_forest = IsolationForest(contamination=0.04, random_state=42)
df['anomaly_label'] = iso_forest.fit_predict(df[['trip_distance', 'fare_amount']])

# Note: scikit-learn labels normal points as 1 and anomalies as -1
normal_data = df[df['anomaly_label'] == 1]
anomaly_data = df[df['anomaly_label'] == -1]

# 3. Create the 2D Scatter Plot
plt.figure(figsize=(10, 6))

# Plot normal instances
plt.scatter(normal_data['trip_distance'], normal_data['fare_amount'],
            c='lightgray', edgecolor='k', alpha=0.7, label='Normal Data')

# Plot detected anomalies
plt.scatter(anomaly_data['trip_distance'], anomaly_data['fare_amount'],
            c='red', edgecolor='k', s=60, marker='x', label='Detected Anomaly')

# Plot design configuration
plt.title('NYC Taxi Fare Outlier Detection via Isolation Forest')
plt.xlabel('Trip Distance (Miles)')
plt.ylabel('Fare Amount ($)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
#Q8. Train a SARIMA model on the monthly airline passengers dataset.
#Forecast the next 12 months and visualize the results.
#(Include your Python code
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm

# 1. Load the complete AirPassengers dataset
passengers = [
    112, 118, 132, 129, 121, 135, 148, 148, 136, 119, 104, 118,
    115, 126, 141, 135, 125, 149, 170, 170, 158, 133, 114, 140,
    145, 150, 178, 163, 172, 178, 199, 199, 184, 162, 146, 166,
    171, 180, 193, 181, 183, 218, 230, 242, 209, 191, 172, 194,
    196, 196, 236, 235, 229, 243, 264, 272, 237, 211, 180, 201,
    204, 188, 235, 227, 234, 264, 302, 293, 259, 229, 203, 229,
    242, 233, 267, 269, 270, 315, 364, 347, 312, 274, 237, 278,
    284, 277, 317, 313, 318, 374, 413, 405, 355, 306, 271, 306,
    315, 301, 356, 348, 355, 422, 465, 467, 404, 347, 305, 336,
    340, 318, 362, 348, 363, 435, 491, 505, 404, 359, 310, 337,
    360, 342, 406, 396, 420, 472, 548, 559, 463, 407, 362, 405,
    417, 391, 419, 461, 472, 535, 622, 606, 508, 461, 390, 432
]

dates = pd.date_range(start='1949-01-01', end='1960-12-01', freq='MS')
df = pd.DataFrame({'Passengers': passengers}, index=dates)

# 2. Fit the SARIMA Model
# Order (p,d,q) tracks the trend; seasonal_order (P,D,Q,s) tracks the 12-month rhythm
# Using standard robust baseline parameters for the AirPassengers series
model = sm.tsa.statespace.SARIMAX(
    df['Passengers'],
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 12),
    enforce_stationarity=False,
    enforce_invertibility=False
)
results = model.fit(disp=False)

# 3. Forecast the next 12 months (calendar year 1961)
forecast_steps = 12
forecast_res = results.get_forecast(steps=forecast_steps)

forecast_index = pd.date_range(start='1961-01-01', periods=forecast_steps, freq='MS')
forecast_mean = forecast_res.predicted_mean
forecast_ci = forecast_res.conf_int()

# 4. Visualization
plt.figure(figsize=(12, 6))

# Plot historical actuals
plt.plot(df.index, df['Passengers'], label='Observed Data', color='blue', linewidth=2)

# Plot point forecast
plt.plot(forecast_index, forecast_mean, label='SARIMA Forecast', color='red', linestyle='--', linewidth=2)

# Plot confidence intervals (shaded uncertainty area)
plt.fill_between(
    forecast_index,
    forecast_ci.iloc[:, 0],
    forecast_ci.iloc[:, 1],
    color='pink',
    alpha=0.4,
    label='95% Confidence Interval'
)

plt.title('AirPassengers 12-Month Out-of-Sample Forecast via SARIMA')
plt.xlabel('Date')
plt.ylabel('Monthly Passengers (Thousands)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

```python
#Q9. Apply Local Outlier Factor (LOF) on any numerical dataset to detect
#anomalies and visualize them using matplotlib.
#(Include your Python code
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neighbors import LocalOutlierFactor

# 1. Synthesize a numerical dataset with dense clusters and isolated outliers
np.random.seed(42)

# Generate two dense clusters (normal patterns)
cluster_1 = np.random.normal(loc=5, scale=1, size=(100, 2))
cluster_2 = np.random.normal(loc=12, scale=1.5, size=(100, 2))

# Generate isolated anomalies (outliers)
anomalies = np.random.uniform(low=0, high=18, size=(15, 2))

# Combine into a unified dataset
X = np.vstack([cluster_1, cluster_2, anomalies])
df = pd.DataFrame(X, columns=['Feature_1', 'Feature_2'])

# 2. Configure and fit the Local Outlier Factor model
# n_neighbors defines the local radius; contamination sets the expected anomaly rate (~7%)
lof = LocalOutlierFactor(n_neighbors=20, contamination=0.07)
df['anomaly_label'] = lof.fit_predict(df[['Feature_1', 'Feature_2']])

# Extract negative outlier scores (lower means more anomalous)
df['negative_outlier_factor'] = lof.negative_outlier_factor_

# Extract groups (LOF returns 1 for inliers and -1 for outliers)
inliers = df[df['anomaly_label'] == 1]
outliers = df[df['anomaly_label'] == -1]

# 3. Create the 2D Scatter Plot
plt.figure(figsize=(10, 6))

# Plot normal inlier points
plt.scatter(inliers['Feature_1'], inliers['Feature_2'],
            c='teal', edgecolor='w', s=40, alpha=0.8, label='Inliers (Normal Data)')

# Plot detected outliers
plt.scatter(outliers['Feature_1'], outliers['Feature_2'],
            c='none', edgecolor='red', s=120, linewidths=2, label='Outliers (Anomalies)')
plt.scatter(outliers['Feature_1'], outliers['Feature_2'],
            c='crimson', marker='x', s=40, label='_nolegend_')

# Plot configurations
plt.title('Anomaly Detection via Local Outlier Factor (LOF)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

```python
#Q10. You are working as a data scientist for a power grid monitoring company.
#Your goal is to forecast energy demand and also detect abnormal spikes or drops in
#real-time consumption data collected every 15 minutes. The dataset includes features
#like timestamp, region, weather conditions, and energy usage.
#Explain your real-time data science workflow:
# How would you detect anomalies in this streaming data (Isolation Forest / LOF /
#DBSCAN)?
# Which time series model would you use for short-term forecasting (ARIMA /
#SARIMA / SARIMAX)?
# How would you validate and monitor the performance over time?
# How would this solution help business decisions or operations?
#(Include your Python code
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from sklearn.ensemble import IsolationForest

# ==========================================
# 1. TELEMETRY STREAM SIMULATION (15-MIN INTERVALS)
# ==========================================
np.random.seed(42)
# Simulate 7 days of 15-minute readings (24h * 4 intervals = 96 steps/day)
n_steps = 96 * 7
timestamps = pd.date_range(start='2026-09-01', periods=n_steps, freq='15min')

# Base drivers: Daily cycle + baseline load growth
base_demand = 150 + 50 * np.sin(2 * np.pi * np.arange(n_steps) / 96)
# Weather driver: Simulated outdoor temperature profile
temperature = 25 + 8 * np.sin(2 * np.pi * np.arange(n_steps) / 96) + np.random.normal(0, 1, n_steps)
# Combine features into historical load profile
energy_usage = base_demand + (temperature - 20) * 4 + np.random.normal(0, 5, n_steps)

# Inject known critical operational faults (Anomalies)
energy_usage[150] += 120  # Extreme surge (e.g., Industrial startup surge)
energy_usage[400] -= 90   # Deep line drop (e.g., Substation failure)

grid_df = pd.DataFrame({
    'timestamp': timestamps,
    'temperature': temperature,
    'energy_usage': energy_usage
}).set_index('timestamp')

# ==========================================
# 2. STREAMING ANOMALY DETECTION (ISOLATION FOREST)
# ==========================================
# Using a 2% expected contamination target for systemic grid failures
iso_forest = IsolationForest(contamination=0.02, random_state=42)
grid_df['anomaly_score'] = iso_forest.fit_predict(grid_df[['temperature', 'energy_usage']])

# Extract anomalous data packets (-1 signifies outlier)
anomalies = grid_df[grid_df['anomaly_score'] == -1]

# ==========================================
# 3. SHORT-TERM DEMAND FORECASTING (SARIMAX)
# ==========================================
# To capture the daily pattern with 15-minute data, seasonal_period=96.
# For faster processing in this simulation, resample to hourly intervals.
hourly_df = grid_df.resample('h').agg({'energy_usage': 'mean', 'temperature': 'mean'})

# Define exogenous parameters and fit weather-dependent SARIMAX(1,1,1)x(1,1,1)_24
sarimax_model = sm.tsa.statespace.SARIMAX(
    hourly_df['energy_usage'],
    exog=hourly_df['temperature'],
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 24),
    enforce_stationarity=False,
    enforce_invertibility=False
)
model_results = sarimax_model.fit(disp=False)

# Forecast the next 24 hours (Requires a weather forecast profile as exog input)
forecast_horizon = 24
future_temp_forecast = 28 + 6 * np.sin(2 * np.pi * np.arange(forecast_horizon) / 24)
forecast_obj = model_results.get_forecast(steps=forecast_horizon, exog=future_temp_forecast)

forecast_index = pd.date_range(start=hourly_df.index[-1] + pd.Timedelta(hours=1), periods=forecast_horizon, freq='h')
forecast_mean = forecast_obj.predicted_mean
forecast_ci = forecast_obj.conf_int()

# ==========================================
# 4. MONITORING DASHBOARD VISUALIZATION
# ==========================================
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10), sharex=False)

# Subplot 1: Real-time telemetry monitoring + anomaly flags
ax1.plot(grid_df.index, grid_df['energy_usage'], color='dodgerblue', label='Grid Consumption', alpha=0.8)
ax1.scatter(anomalies.index, anomalies['energy_usage'], color='crimson', marker='X', s=100, zorder=5, label='Grid Anomaly Flag')
ax1.set_title('Real-Time Power Grid Telemetry & Anomaly Alerts')
ax1.set_ylabel('Load (MW)')
ax1.grid(True, linestyle='--', alpha=0.5)
ax1.legend(loc='upper left')

# Subplot 2: Weather-adjusted out-of-sample forecast
ax2.plot(hourly_df.index[-72:], hourly_df['energy_usage'].tail(72), label='Recent Historical Load', color='black')
ax2.plot(forecast_index, forecast_mean, label='Weather-Adjusted SARIMAX Forecast', color='orange', linestyle='--', linewidth=2)
ax2.fill_between(forecast_index, forecast_ci.iloc[:, 0], forecast_ci.iloc[:, 1], color='orange', alpha=0.2, label='95% Predictive Guardrail')
ax2.set_title('Short-Term Operational Load Forecasting (Next 24 Hours)')
ax2.set_ylabel('Load (MW)')
ax2.grid(True, linestyle='--', alpha=0.5)
ax2.legend(loc='upper left')

plt.tight_layout()
plt.show()
```
